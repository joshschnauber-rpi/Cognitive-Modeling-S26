```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from cmdstanpy import CmdStanModel

import arviz as az
```

```python
# Get data
data = pd.read_csv('data/recognition.csv', header=0, sep=',')
data.head()

# Prepare data in Stan friendly format
data_dict = {
    'N': data.shape[0],
    'old': data.old.values,
    'response': data.response.values
}
```

```python
# Compile 1HT model
model1 = CmdStanModel(stan_file="hw4_2_1.stan")
```

```python
# Sample 1HT model
fit1 = model1.sample(data=data_dict, chains=4, iter_sampling=2500, iter_warmup=1000)

# Display sampling diagnostics
print(fit1.diagnose())
```

```python
# Compile 2HT model
model2 = CmdStanModel(stan_file="hw4_2_2.stan")
```

```python
# Sample 2HT model
fit2 = model2.sample(data=data_dict, chains=4, iter_sampling=2500, iter_warmup=1000)

# Display sampling diagnostics
print(fit2.diagnose())
```

```python
# Display 1HT sampling data
summary = fit1.summary()
display(summary)

draws_df = fit1.draws_pd()
display(draws_df.head())
```

```python
# Display 2HT sampling data
summary = fit2.summary()
display(summary)

draws_df = fit2.draws_pd()
display(draws_df.head())
```

The One-High-Threshold model estimates `d` with a mean of `0.675` and `g` with a mean of `0.347`.
The Two-High-Threshold model estimates `d` with a mean of `0.673` and `g` with a mean of `0.547`.

The same `d` value implies that both models have the same confidence in selecting
the correct answer. But, the 1HT model having a lower `g` value implies that the model, when unsure of the correct answer, will be more likely to respond that the word is new.
The `g` value for the 2HT model implies that the model has more of a 50/50 chance of
picking either reponse when unsure.

This most likely occurs because the 1HT model assumes that when looking at a new word, 
the person responding will be unsure of their guess. But, that isn't necessarily the case,
so if the data shows that new words were correctly labeled as new more often than old,
then it would push the value of `g` down to account for that.

But the 2HT model does not treat all responses to new words as guesses, so the actual
guessing part is more of a 50/50.
