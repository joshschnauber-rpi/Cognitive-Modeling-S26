```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from cmdstanpy import CmdStanModel

import arviz as az
```

```python
# Get data
data = pd.read_csv('data/recognition.csv', header=0, sep=',')
data.head()

# Prepare data in Stan friendly format
data_dict = {
    'N': data.shape[0],
    'old': data.old.values,
    'response': data.response.values
}
```

```python
# Compile 1HT model
model1 = CmdStanModel(stan_file="hw4_3_1.stan")
```

```python
# Sample 1HT model
fit1 = model1.sample(data=data_dict, chains=4, iter_sampling=2500, iter_warmup=1000)

# Display sampling diagnostics
print(fit1.diagnose())
```

```python
# Compile 2HT model
model2 = CmdStanModel(stan_file="hw4_3_2.stan")
```

```python
# Sample 2HT model
fit2 = model2.sample(data=data_dict, chains=4, iter_sampling=2500, iter_warmup=1000)

# Display sampling diagnostics
print(fit2.diagnose())
```

```python
# Display 1HT sampling data
summary = fit1.summary()
display(summary)

draws_df = fit1.draws_pd()
display(draws_df.head())

# Plot
data1 = az.from_cmdstanpy(posterior=fit1)
_ = az.plot_trace(data1, legend=True, var_names=["d", "g"])
plt.tight_layout()
```

```python
# Display 2HT sampling data
summary = fit2.summary()
display(summary)

draws_df = fit2.draws_pd()
display(draws_df.head())

# Plot
_ = az.plot_trace(fit2, legend=True)
plt.tight_layout()

# Display sampling diagnostics
print(fit2.diagnose())
```
